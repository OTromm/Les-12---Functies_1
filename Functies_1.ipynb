Als we veel code hebben staan, wordt de code snel onoverzichtelijk en slordig. Om meer structuur te brengen in onze code maken we gebruik van `functies`. `Functies` zijn blokken waarin we code kunnen schrijven en deze een naam geven zodat we al deze regels code kunnen herhalen zonder alles opnieuw te typen. Als we een `functie` aanmaken met het sleutelwoord `def`, wordt deze opgeslagen maar niet uitgevoerd. Pas als we de naam van de `functie` in de code gebruiken, dan wordt de code in de `functie` uitgevoerd.

```python
def begroet():
    print("Hallo daar")
    print("Het is lekker weer vandaag")

begroet()
begroet()
```

## Parameters

Soms heeft een `functie` informatie nodig om te werken, en deze informatie kan verschillen elke keer als we de `functie` aanroepen. Hiervoor maken we gebruik van `parameters`. Tijdens het definiëren van de `functie` worden er `variabelen` aangemaakt. Deze krijgen pas een waarde als de `functie` gebruikt wordt. De waardes worden tijdens het aanroepen van de `functie` tussen de `()` van de naam van de `functie` gezet.

```python
def begroet(naam):
    print("Hallo " + naam)

begroet("Charlotte")
begroet("Jurre")
```

## Teruggeven

We kunnen ook `functies` gebruiken om waardes voor ons te verwerken die we daarna weer in onze code kunnen gebruiken. Een `functie` kan eindigen met een `return` zodat er iets wordt teruggegeven. Dit kunnen we vervolgens weer opslaan in een variabelen.

```python
def verdubbel(getal):
    return getal * 2

x = 5
y = verdubbel(x)
print(y)
```
